```python
from pathlib import Path
import zipfile

download_dir = (
    Path.home()
    / "tensorflow_datasets"
    / "downloads"
    / "plant_village"
)


archives = [
    f for f in download_dir.iterdir()
    if f.is_file() and not f.name.endswith(".INFO")
]

if not archives:
    raise FileNotFoundError("PlantVillage archive not found.")

archive_path = max(archives, key=lambda f: f.stat().st_size)


extract_dir = Path.home() / "PlantDiseaseDetection" / "dataset"

extract_dir.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(archive_path, "r") as archive:

    
    root = extract_dir.resolve()

    for member in archive.infolist():
        target = (extract_dir / member.filename).resolve()

        if not target.is_relative_to(root):
            raise ValueError(
                f"Unsafe archive path detected: {member.filename}"
            )

    print("Extracting PlantVillage dataset...")
    archive.extractall(extract_dir)

print("\nExtraction completed!")
print("Dataset location:", extract_dir)
```

```python
from pathlib import Path
import pandas as pd
import tensorflow_datasets as tfds

# Original extracted dataset
DATASET_DIR = (
    Path.home()
    / "PlantDiseaseDetection"
    / "dataset"
)

# Actual image dataset root
IMAGE_ROOT = (
    DATASET_DIR
    / "Plant_leave_diseases_dataset_without_augmentation"
)

# Official 38 class names
builder = tfds.builder("plant_village")

EXPECTED_CLASSES = builder.info.features["label"].names

IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp"}

records = []

for class_name in EXPECTED_CLASSES:

    class_dir = IMAGE_ROOT / class_name

    if not class_dir.is_dir():
        print("WARNING: Missing class directory:", class_name)
        continue

    for image_path in class_dir.iterdir():

        if (
            image_path.is_file()
            and image_path.suffix.lower() in IMAGE_EXTENSIONS
        ):
            records.append({
                "image_path": str(image_path),
                "class_name": class_name
            })

# Create DataFrame
df = pd.DataFrame(records)

if df.empty:
    raise ValueError(
        "No images found. Check IMAGE_ROOT and class folder names."
    )

print("=" * 60)
print("CLEAN DATASET SUMMARY")
print("=" * 60)

print("Total images:", f"{len(df):,}")
print("Unique classes:", df["class_name"].nunique())

print("\nImages per class:")
display(
    df["class_name"]
    .value_counts()
    .sort_index()
    .to_frame("image_count")
)

# Save the index
PROJECT_DIR = Path.home() / "PlantDiseaseDetection"
index_path = PROJECT_DIR / "image_index.csv"

df.to_csv(index_path, index=False)

print("\nDataset index saved to:")
print(index_path)
```

```python
import tensorflow as tf
import pandas as pd
import numpy as np
from pathlib import Path


PROJECT_DIR = Path.home() / "PlantDiseaseDetection"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE


train_df = pd.read_csv(PROJECT_DIR / "train.csv")
val_df = pd.read_csv(PROJECT_DIR / "validation.csv")
test_df = pd.read_csv(PROJECT_DIR / "test.csv")


class_names = sorted(train_df["class_name"].unique())

class_to_index = {
    class_name: index
    for index, class_name in enumerate(class_names)
}

NUM_CLASSES = len(class_names)

print("Number of classes:", NUM_CLASSES)
print("Image size:", IMAGE_SIZE)
print("Batch size:", BATCH_SIZE)


train_labels = train_df["class_name"].map(class_to_index).to_numpy()
val_labels = val_df["class_name"].map(class_to_index).to_numpy()
test_labels = test_df["class_name"].map(class_to_index).to_numpy()


assert not pd.isna(train_labels).any()
assert not pd.isna(val_labels).any()
assert not pd.isna(test_labels).any()

print("\nLabel mapping verified.")
```

```python
def load_image(image_path, label):
    """
    Load an image, convert it to RGB, resize it,
    and normalize pixel values to the range [0, 1].
    """

    image = tf.io.read_file(image_path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    image = tf.cast(image, tf.float32) / 255.0

    return image, label


def create_dataset(dataframe, labels, training=False):

    dataset = tf.data.Dataset.from_tensor_slices(
        (
            dataframe["image_path"].astype(str).to_numpy(),
            labels.astype(np.int32)
        )
    )

    if training:
        dataset = dataset.shuffle(
            buffer_size=min(len(dataframe), 5000),
            seed=42,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=AUTOTUNE
    )

    dataset = dataset.batch(BATCH_SIZE)

    dataset = dataset.prefetch(AUTOTUNE)

    return dataset



train_ds = create_dataset(
    train_df,
    train_labels,
    training=True
)

val_ds = create_dataset(
    val_df,
    val_labels
)

test_ds = create_dataset(
    test_df,
    test_labels
)

print("Training, validation, and testing pipelines created.")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models


tf.keras.utils.set_random_seed(42)


cnn_model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    
    layers.Conv2D(
        32, (3, 3),
        padding="same",
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    
    layers.Conv2D(
        64, (3, 3),
        padding="same",
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    #
    layers.Conv2D(
        128, (3, 3),
        padding="same",
        activation="relu"
    ),
    layers.MaxPooling2D((2, 2)),

    
    layers.GlobalAveragePooling2D(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.4),

    
    layers.Dense(NUM_CLASSES, activation="softmax")
])


cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


cnn_model.summary()
```

```python
import tensorflow as tf
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT_DIR = Path.home() / "PlantDiseaseDetection"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16
AUTOTUNE = tf.data.AUTOTUNE


train_df = pd.read_csv(PROJECT_DIR / "train.csv")
val_df = pd.read_csv(PROJECT_DIR / "validation.csv")
test_df = pd.read_csv(PROJECT_DIR / "test.csv")


class_names = sorted(train_df["class_name"].unique())

class_to_index = {
    name: index
    for index, name in enumerate(class_names)
}

NUM_CLASSES = len(class_names)

train_labels = train_df["class_name"].map(class_to_index).to_numpy()
val_labels = val_df["class_name"].map(class_to_index).to_numpy()
test_labels = test_df["class_name"].map(class_to_index).to_numpy()


def load_image_mobilenet(image_path, label):
    """Load and preprocess an image for MobileNetV2."""

    image = tf.io.read_file(image_path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    image = tf.cast(image, tf.float32)

    
    image = tf.keras.applications.mobilenet_v2.preprocess_input(
        image
    )

    return image, label


def create_mobilenet_dataset(dataframe, labels, training=False):

    ds = tf.data.Dataset.from_tensor_slices(
        (
            dataframe["image_path"].astype(str).to_numpy(),
            labels.astype(np.int32)
        )
    )

    if training:
        ds = ds.shuffle(
            buffer_size=min(len(dataframe), 5000),
            seed=42,
            reshuffle_each_iteration=True
        )

    ds = ds.map(
        load_image_mobilenet,
        num_parallel_calls=AUTOTUNE
    )

    ds = ds.batch(BATCH_SIZE)
    ds = ds.prefetch(AUTOTUNE)

    return ds


train_ds_mobilenet = create_mobilenet_dataset(
    train_df,
    train_labels,
    training=True
)

val_ds_mobilenet = create_mobilenet_dataset(
    val_df,
    val_labels
)

test_ds_mobilenet = create_mobilenet_dataset(
    test_df,
    test_labels
)

print("MobileNetV2 data pipeline created.")
print("Number of classes:", NUM_CLASSES)
print("Image size:", IMAGE_SIZE)
print("Batch size:", BATCH_SIZE)
```

```python
for images, labels in train_ds_mobilenet.take(1):

    print("Image batch shape:", images.shape)
    print("Label batch shape:", labels.shape)

    print(
        "Pixel range:",
        float(tf.reduce_min(images).numpy()),
        "to",
        float(tf.reduce_max(images).numpy())
    )

    print("First label:", int(labels[0].numpy()))
    print("Class:", class_names[int(labels[0].numpy())])
```

```python
from tensorflow.keras import layers, models

tf.keras.utils.set_random_seed(42)


base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights="imagenet"
)


base_model.trainable = False


inputs = layers.Input(shape=(224, 224, 3))

x = base_model(inputs, training=False)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

mobilenet_model = models.Model(
    inputs=inputs,
    outputs=outputs
)

mobilenet_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

mobilenet_model.summary()
```

```python
import time

callbacks_mobilenet = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(
            PROJECT_DIR / "best_mobilenetv2.keras"
        ),
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    )
]

print("Starting MobileNetV2 training...")

start_time = time.time()

mobilenet_history = mobilenet_model.fit(
    train_ds_mobilenet,
    validation_data=val_ds_mobilenet,
    epochs=5,
    callbacks=callbacks_mobilenet,
    verbose=1
)

elapsed_time = time.time() - start_time

print(
    f"\nTraining completed in {elapsed_time / 60:.2f} minutes."
)
```

```python
import matplotlib.pyplot as plt

history = mobilenet_history.history

epochs_range = range(1, len(history["loss"]) + 1)

plt.figure(figsize=(8, 5))

plt.plot(
    epochs_range,
    history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    epochs_range,
    history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.title("MobileNetV2 - Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))

plt.plot(
    epochs_range,
    history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    epochs_range,
    history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.title("MobileNetV2 - Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

PROJECT_DIR = Path.home() / "PlantDiseaseDetection"

MODEL_PATH = PROJECT_DIR / "best_mobilenetv2.keras"
LABELS_PATH = PROJECT_DIR / "class_names.json"
TEST_PATH = PROJECT_DIR / "test.csv"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16


model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)

with open(LABELS_PATH, "r", encoding="utf-8") as file:
    class_names = json.load(file)

test_df = pd.read_csv(TEST_PATH)

class_to_index = {
    name: index
    for index, name in enumerate(class_names)
}

y_true = test_df["class_name"].map(
    class_to_index
).to_numpy(dtype=np.int32)


def load_test_image(image_path, label):
    image = tf.io.read_file(image_path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    image = tf.cast(image, tf.float32)

    image = tf.keras.applications.mobilenet_v2.preprocess_input(
        image
    )

    return image, label


test_ds = tf.data.Dataset.from_tensor_slices(
    (
        test_df["image_path"].astype(str).to_numpy(),
        y_true
    )
)

test_ds = test_ds.map(
    load_test_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.batch(BATCH_SIZE).prefetch(
    tf.data.AUTOTUNE
)

print("Evaluating deployment model...")
print("Test images:", len(test_df))

# Generate predictions
y_prob = model.predict(
    test_ds,
    verbose=1
)

y_pred = np.argmax(y_prob, axis=1)

# Calculate metrics
test_accuracy = accuracy_score(y_true, y_pred)

macro_f1 = f1_score(
    y_true,
    y_pred,
    average="macro",
    zero_division=0
)

weighted_f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 60)
print("DEPLOYMENT MODEL — TEST RESULTS")
print("=" * 60)

print(f"Test accuracy:    {test_accuracy:.4%}")
print(f"Macro F1-score:   {macro_f1:.4f}")
print(f"Weighted F1-score:{weighted_f1:.4f}")


report = classification_report(
    y_true,
    y_pred,
    labels=np.arange(len(class_names)),
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

print("\n10 classes with the lowest F1-scores:")

display(
    report_df.loc[
        class_names,
        ["precision", "recall", "f1-score", "support"]
    ]
    .sort_values("f1-score")
    .head(10)
)
```

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.metrics import confusion_matrix

PROJECT_DIR = Path.home() / "PlantDiseaseDetection"

MODEL_PATH = PROJECT_DIR / "best_mobilenetv2.keras"
LABELS_PATH = PROJECT_DIR / "class_names.json"
TEST_PATH = PROJECT_DIR / "test.csv"

IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16


model = tf.keras.models.load_model(
    MODEL_PATH,
    compile=False
)


with open(LABELS_PATH, "r", encoding="utf-8") as file:
    class_names = json.load(file)


test_df = pd.read_csv(TEST_PATH).reset_index(drop=True)

class_to_index = {
    name: index
    for index, name in enumerate(class_names)
}


unknown_classes = (
    set(test_df["class_name"].unique())
    - set(class_names)
)

if unknown_classes:
    raise ValueError(
        f"Unknown classes found: {unknown_classes}"
    )

y_true = test_df["class_name"].map(
    class_to_index
).to_numpy(dtype=np.int32)


def load_test_image(image_path, label):
    image = tf.io.read_file(image_path)

    image = tf.io.decode_image(
        image,
        channels=3,
        expand_animations=False
    )

    image.set_shape([None, None, 3])

    image = tf.image.resize(
        image,
        IMAGE_SIZE
    )

    image = tf.cast(image, tf.float32)

    image = tf.keras.applications.mobilenet_v2.preprocess_input(
        image
    )

    return image, label


test_ds = tf.data.Dataset.from_tensor_slices(
    (
        test_df["image_path"].astype(str).to_numpy(),
        y_true
    )
)

test_ds = test_ds.map(
    load_test_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.batch(
    BATCH_SIZE
).prefetch(
    tf.data.AUTOTUNE
)

print("Model loaded:", MODEL_PATH.name)
print("Test images:", len(test_df))
print("Disease classes:", len(class_names))
print("Dataset prepared successfully.")
```

```python
print("Generating predictions for the test set...")

y_prob = model.predict(
    test_ds,
    verbose=1
)

y_pred = np.argmax(
    y_prob,
    axis=1
)

confidence = np.max(
    y_prob,
    axis=1
)


results_df = test_df.copy()

results_df["true_index"] = y_true
results_df["predicted_index"] = y_pred

results_df["true_class"] = [
    class_names[i] for i in y_true
]

results_df["predicted_class"] = [
    class_names[i] for i in y_pred
]

results_df["confidence"] = confidence

results_df["correct"] = (
    results_df["true_index"]
    == results_df["predicted_index"]
)


errors_df = results_df.loc[
    ~results_df["correct"]
].copy()

print("\n" + "=" * 55)
print("MISCLASSIFICATION SUMMARY")
print("=" * 55)

print("Total test images:", len(results_df))
print("Correct predictions:", int(results_df["correct"].sum()))
print("Incorrect predictions:", len(errors_df))

print(
    "Calculated accuracy:",
    f"{results_df['correct'].mean():.2%}"
)

print("\nFirst 15 incorrect predictions:")

display(
    errors_df[
        [
            "image_path",
            "true_class",
            "predicted_class",
            "confidence"
        ]
    ].head(15)
)
```

```python
confusion = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(len(class_names))
)

confusion_df = pd.DataFrame(
    confusion,
    index=class_names,
    columns=class_names
)


np.fill_diagonal(confusion, 0)

confusion_pairs = []

for true_idx, pred_idx in zip(
    *np.where(confusion > 0)
):
    confusion_pairs.append({
        "Actual class": class_names[true_idx],
        "Predicted class": class_names[pred_idx],
        "Number of mistakes": int(
            confusion[true_idx, pred_idx]
        )
    })

confusion_pairs_df = pd.DataFrame(
    confusion_pairs,
    columns=[
        "Actual class",
        "Predicted class",
        "Number of mistakes"
    ]
)

confusion_pairs_df = confusion_pairs_df.sort_values(
    "Number of mistakes",
    ascending=False
)

print("Top 15 most frequent classification errors:")

display(
    confusion_pairs_df.head(15)
)


errors_df.to_csv(
    PROJECT_DIR / "misclassified_images.csv",
    index=False
)

confusion_pairs_df.to_csv(
    PROJECT_DIR / "confusion_pairs.csv",
    index=False
)

print("\nAnalysis files saved:")
print(PROJECT_DIR / "misclassified_images.csv")
print(PROJECT_DIR / "confusion_pairs.csv")
```

```python
import pandas as pd


class_error_summary = (
    results_df.groupby("true_class")
    .agg(
        Total_Images=("correct", "size"),
        Correct_Predictions=("correct", "sum")
    )
)

class_error_summary["Incorrect_Predictions"] = (
    class_error_summary["Total_Images"]
    - class_error_summary["Correct_Predictions"]
)

class_error_summary["Accuracy"] = (
    class_error_summary["Correct_Predictions"]
    / class_error_summary["Total_Images"]
)

class_error_summary["Error_Rate"] = (
    class_error_summary["Incorrect_Predictions"]
    / class_error_summary["Total_Images"]
)

class_error_summary = class_error_summary.sort_values(
    "Error_Rate",
    ascending=False
)

print("10 classes with the highest error rates:")

display(
    class_error_summary.head(10).style.format({
        "Accuracy": "{:.2%}",
        "Error_Rate": "{:.2%}"
    })
)


summary_path = PROJECT_DIR / "class_error_summary.csv"

class_error_summary.to_csv(summary_path)

print("\nSaved:", summary_path)
```

```python
from pathlib import Path
import json
import numpy as np
import tensorflow as tf
from PIL import Image

PROJECT_DIR = Path.home() / "PlantDiseaseDetection"

ORIGINAL_MODEL_PATH = PROJECT_DIR / "best_mobilenetv2.keras"
FINAL_MODEL_PATH = PROJECT_DIR / "final_plant_disease_model.keras"
LABELS_PATH = PROJECT_DIR / "class_names.json"


IMAGE_PATH = Path(r"C:\Users\thesh\PlantDiseaseDetection\dataset\Plant_leave_diseases_dataset_without_augmentation\Apple___healthy\image (3).JPG")

with open(LABELS_PATH, "r", encoding="utf-8") as f:
    class_names = json.load(f)

original_model = tf.keras.models.load_model(
    ORIGINAL_MODEL_PATH,
    compile=False
)

final_model = tf.keras.models.load_model(
    FINAL_MODEL_PATH,
    compile=False
)


image = Image.open(IMAGE_PATH).convert("RGB")
image = image.resize((224, 224))

image_array = np.asarray(
    image,
    dtype=np.float32
)

image_array = tf.keras.applications.mobilenet_v2.preprocess_input(
    image_array
)

image_array = np.expand_dims(image_array, axis=0)

# P
original_probabilities = original_model.predict(
    image_array,
    verbose=0
)[0]

final_probabilities = final_model.predict(
    image_array,
    verbose=0
)[0]

print("ORIGINAL MODEL")
print(
    class_names[int(np.argmax(original_probabilities))]
)
print(
    f"Confidence: {np.max(original_probabilities):.2%}"
)

print("\nFINE-TUNED MODEL")
print(
    class_names[int(np.argmax(final_probabilities))]
)
print(
    f"Confidence: {np.max(final_probabilities):.2%}"
)

print("\nTop 3 predictions from the fine-tuned model:")

top_indices = np.argsort(final_probabilities)[-3:][::-1]

for index in top_indices:
    print(
        f"{class_names[index]}: "
        f"{final_probabilities[index]:.2%}"
    )
```
